# Step 4: Data Preparation and Feature Engineering

Before conducting exploratory analysis, several variables require transformation into more suitable formats.

The original dataset stores dates and times as text variables. Converting these variables into datetime objects enables temporal analysis and simplifies the extraction of useful features such as month and hour.

Additional variables are also created to facilitate the investigation of accident patterns across different periods of the day and seasons.

The main objectives of this preprocessing stage are:

- Convert the date variable into datetime format.
- Extract month information.
- Extract hour information.
- Create a categorical time-of-day variable.
- Verify that all transformations have been applied correctly.

## STEP 4.1 : Convert Date Column



```python
# Convert the date column into datetime format
df['date'] = pd.to_datetime(df['date'], dayfirst=True)

# Display data type
print(df['date'].dtype)
```

## STEP 4.2 : Extract Month and Hour

```python
# Create month column
df['month'] = df['date'].dt.month

# Convert time column to string
df['time'] = df['time'].astype(str)

# Extract hour from time
df['hour'] = df['time'].str[:2].astype(int)

# Display first five rows
df[['date','month','time','hour']].head()
```

## STEP 4.3 : Create Time of Day Categories

```python
def classify_time(hour):
    if 5 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    else:
        return "Night"

# Apply the function
df['time_period'] = df['hour'].apply(classify_time)

# Check the results
df['time_period'].value_counts()
```

## Observations

The preprocessing stage successfully transformed the temporal variables into formats suitable for analysis.

Two new variables were created:

- **Month**, allowing seasonal trend analysis.
- **Time Period**, enabling the investigation of accidents occurring during the morning, afternoon, evening, and night.

These engineered features provide additional analytical value and improve the interpretability of subsequent visualizations.

## Visualization 1: Distribution of Accident Severity

Understanding the severity distribution is essential for identifying the overall impact of road traffic collisions.

The UK Department for Transport classifies collisions into three severity levels:

- 1 = Fatal
- 2 = Serious
- 3 = Slight

This visualization explores how accident severity is distributed across all recorded collisions in 2024.

A bar chart is used because it provides a simple and effective comparison between discrete categories.

```python
# Count accident severity categories
severity_counts = df['collision_severity'].value_counts().sort_index()

# Create figure
plt.figure(figsize=(8,6))

# Plot bar chart
sns.barplot(
    x=severity_counts.index,
    y=severity_counts.values,
    palette="viridis"
)

# Add labels
plt.title("Distribution of Road Accident Severity", fontsize=15)
plt.xlabel("Collision Severity")
plt.ylabel("Number of Accidents")

# Replace numeric labels
plt.xticks(
    ticks=[0,1,2],
    labels=["Fatal (1)", "Serious (2)", "Slight (3)"]
)

# Add values above bars
for i, value in enumerate(severity_counts.values):
    plt.text(i, value + 500,
             str(value),
             ha='center',
             fontsize=10)

plt.show()
```

# Visualization 4: Accident Severity by Time of Day

While the previous visualization examined the overall distribution of accidents across different periods of the day, it is equally important to investigate whether accident severity varies depending on when collisions occur.

This analysis combines the engineered **time_period** variable with the official **collision_severity** classification to identify potential temporal patterns in accident outcomes.

A grouped bar chart is used because it allows multiple categories to be compared simultaneously across different time periods.

```python
# Create contingency table
severity_time = pd.crosstab(
    df['time_period'],
    df['collision_severity']
)

# Arrange rows in logical order
severity_time = severity_time.reindex(
    ['Morning', 'Afternoon', 'Evening', 'Night']
)

# Rename columns
severity_time.columns = ['Fatal', 'Serious', 'Slight']

plt.figure(figsize=(12,6), dpi=120)

ax = severity_time.plot(
    kind='bar',
    width=0.8,
    color=['#C0392B', '#2980B9', '#27AE60'],  # Red, Blue, Green
    figsize=(12,6)
)

plt.title(
    "Accident Severity Across Different Times of Day",
    fontsize=16,
    weight='bold',
    pad=15
)

plt.xlabel("Time Period", fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)

plt.xticks(rotation=0, fontsize=11)
plt.yticks(fontsize=10)

plt.legend(
    title="Severity",
    fontsize=11,
    title_fontsize=12
)

plt.grid(axis='y', linestyle='--', alpha=0.3)

plt.tight_layout()
plt.show()
```

```python
severity_time
```

## Interpretation

The grouped bar chart compares the distribution of collision severity across different periods of the day.

Although **Slight** collisions dominate every time category, the overall number of accidents is highest during the **Afternoon**, reflecting increased traffic activity during working hours and school commuting periods.

The Morning and Evening periods also record substantial numbers of collisions, whereas Night-time accidents occur less frequently overall.

Despite having the lowest overall collision frequency, Night-time accidents appear to have a relatively higher proportion of fatal and serious outcomes compared with other periods of the day. This may be associated with factors such as reduced visibility, driver fatigue, and impaired driving conditions.

These findings suggest that both traffic volume and environmental factors contribute to accident severity patterns throughout the day.

# Visualization 6: Accident Severity by Weather Conditions

Weather conditions can influence road visibility, driver behaviour, braking distance and overall collision risk. For this reason, it is useful to investigate whether accident severity varies under different weather conditions.

This analysis uses the official `weather_conditions` variable and compares it with `collision_severity`. A grouped bar chart is used because it allows the distribution of Fatal, Serious and Slight collisions to be compared across weather categories.

```python
# Create a copy of the dataset
weather_df = df.copy()

# Map weather condition codes to readable labels
weather_labels = {
    1: "Fine",
    2: "Rain",
    3: "Snow",
    4: "Fine + Wind",
    5: "Rain + Wind",
    6: "Snow + Wind",
    7: "Fog/Mist",
    8: "Other",
    9: "Unknown"
}

# Map collision severity codes to labels
severity_labels = {
    1: "Fatal",
    2: "Serious",
    3: "Slight"
}

# Create readable category labels
weather_df["weather_label"] = weather_df["weather_conditions"].map(weather_labels)
weather_df["severity_label"] = weather_df["collision_severity"].map(severity_labels)

# Create contingency table
weather_severity = pd.crosstab(
    weather_df["weather_label"],
    weather_df["severity_label"]
)

# Arrange severity columns
weather_severity = weather_severity[
    ["Fatal", "Serious", "Slight"]
]

# Sort weather conditions by total number of accidents
weather_severity["Total"] = weather_severity.sum(axis=1)
weather_severity = weather_severity.sort_values(
    "Total",
    ascending=False
)
weather_severity = weather_severity.drop(columns="Total")

# Display summary table
weather_severity
```

```python
# Create grouped bar chart
ax = weather_severity.plot(
    kind='bar',
    figsize=(14,7),
    width=0.65,   # instead of 0.75
    color=['#C0392B', '#2980B9', '#27AE60'],
    edgecolor='black',
    linewidth=0.4
)

# Chart title and labels
plt.title(
    "Accident Severity Across Weather Conditions",
    fontsize=16,
    weight="bold",
    pad=15
)

plt.xlabel("Weather Condition", fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)

# Improve axis formatting
plt.xticks(rotation=15, ha="right", fontsize=10)
plt.yticks(fontsize=10)

# Add gridlines
plt.grid(axis="y", linestyle="--", alpha=0.3)

# Add legend
plt.legend(
    title="Severity",
    fontsize=11,
    title_fontsize=12,
    loc="upper right",
    frameon=True
)

# Adjust layout
plt.tight_layout()

# Display chart
plt.show()
```

### Interpretation

The grouped bar chart shows that the majority of recorded collisions occurred during **fine weather conditions**, with Slight accidents representing the largest category.

Rainy weather also records a considerable number of collisions, although substantially fewer than under fine weather conditions.

Although adverse weather conditions record lower overall accident frequencies, they may still increase the likelihood of more severe outcomes due to reduced visibility, longer braking distances, and more challenging driving conditions.

Overall, these findings suggest that weather conditions should be considered alongside traffic exposure, as a lower number of accidents does not necessarily indicate lower risk.

# Visualization 9: Accident Severity by Junction Control

Junction control is an important road infrastructure factor because it influences how drivers behave at intersections. Different junction controls, such as traffic signals, give-way controls, stop signs, and uncontrolled junctions, may be associated with different accident patterns.

This analysis investigates whether accident severity varies across different junction control categories. The official `junction_control` variable is compared with `collision_severity`.

A grouped bar chart is used because it allows Fatal, Serious, and Slight collisions to be compared across each junction control type.

```python
# Create a copy of the dataset for this analysis
junction_df = df.copy()

# Map junction control codes to readable labels
junction_labels = {
    -1: "Data missing",
     0: "Not at junction",
     1: "Authorised person",
     2: "Auto traffic signal",
     3: "Stop sign",
     4: "Give way/uncontrolled"
}

# Map collision severity codes to readable labels
severity_labels = {
    1: "Fatal",
    2: "Serious",
    3: "Slight"
}

# Create readable category labels
junction_df["junction_label"] = junction_df["junction_control"].map(junction_labels)
junction_df["severity_label"] = junction_df["collision_severity"].map(severity_labels)

# Create contingency table
junction_severity = pd.crosstab(
    junction_df["junction_label"],
    junction_df["severity_label"]
)

# Arrange severity columns in logical order
junction_severity = junction_severity[["Fatal", "Serious", "Slight"]]

# Sort categories by total number of accidents
junction_severity["Total"] = junction_severity.sum(axis=1)
junction_severity = junction_severity.sort_values("Total", ascending=False)
junction_severity = junction_severity.drop(columns="Total")

# Display summary table
junction_severity
```

```python
ax = junction_severity.plot(
    kind="bar",
    figsize=(14, 7),
    width=0.65,
    color=["#C0392B", "#2980B9", "#27AE60"],
    edgecolor="black",
    linewidth=0.4
)

# Chart title and labels
plt.title(
    "Accident Severity Across Junction Control Types",
    fontsize=16,
    weight="bold",
    pad=15
)

plt.xlabel("Junction Control Type", fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)

# Improve axis readability
plt.xticks(rotation=20, ha="right", fontsize=10)
plt.yticks(fontsize=10)

# Add light gridlines
plt.grid(axis="y", linestyle="--", alpha=0.3)

# Add legend
plt.legend(
    title="Severity",
    fontsize=11,
    title_fontsize=12,
    loc="upper right",
    frameon=True
)

plt.tight_layout()
plt.show()
```

## Interpretation

The grouped bar chart compares the distribution of Fatal, Serious, and Slight accidents across different junction control types.

The results show that the highest number of recorded collisions occur at junctions controlled by **give-way or uncontrolled arrangements**, with Slight accidents representing the largest category. This may reflect the high frequency of these junction types across the road network and the greater number of driver decision points involved.

Accidents not occurring at junctions also account for a substantial number of cases, suggesting that collision risk is not limited to intersection environments. Controlled junctions such as traffic signals and stop signs record fewer accidents overall, although they still include Serious and Fatal outcomes.

Overall, these findings suggest that junction control should be considered as an important infrastructure factor in road safety analysis. However, the results should be interpreted alongside traffic exposure, junction design, road layout, and driver behaviour, as lower accident counts do not automatically indicate safer conditions.

# Visualization 11: Accident Severity by Day of the Week

The day on which a collision occurs may influence accident frequency and severity because traffic volumes and travel behaviour vary throughout the week. Weekdays often experience heavy commuter traffic, while weekends may involve different driving patterns and recreational travel.

This analysis investigates whether accident severity differs across the days of the week. The official `day_of_week` variable is compared with `collision_severity`.

A line and area chart is used because the day-of-week variable is ordered and helps show weekly accident patterns clearly.

```python
day_df = df.copy()

day_labels = {
    1: "Sunday",
    2: "Monday",
    3: "Tuesday",
    4: "Wednesday",
    5: "Thursday",
    6: "Friday",
    7: "Saturday"
}

day_df["day_label"] = day_df["day_of_week"].map(day_labels)

day_counts = day_df["day_label"].value_counts()

day_order = ["Sunday", "Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday"]
day_counts = day_counts.reindex(day_order)

plt.figure(figsize=(14, 6))

plt.plot(
    day_counts.index,
    day_counts.values,
    marker="o",
    linewidth=3,
    markersize=8,
    color="#8E44AD"
)

plt.fill_between(
    day_counts.index,
    day_counts.values,
    alpha=0.15,
    color="#8E44AD"
)

plt.title("Weekly Trend of Road Traffic Collisions", fontsize=16, weight="bold", pad=15)
plt.xlabel("Day of Week", fontsize=12)
plt.ylabel("Number of Accidents", fontsize=12)
plt.grid(axis="y", linestyle="--", alpha=0.3)

plt.tight_layout()
plt.show()
```

## Interpretation

The grouped bar chart compares the distribution of Fatal, Serious, and Slight accidents across the days of the week.

The results show that accident frequencies vary throughout the week, with Slight accidents representing the largest category on every day. Weekdays often record a high number of collisions, which may reflect increased traffic volumes associated with commuting and business activities.

Weekend collisions also contribute substantially to the overall total and include a notable number of Serious and Fatal accidents. Differences in travel behaviour, recreational journeys, and traffic patterns may influence these outcomes.

Overall, these findings suggest that the timing of road travel should be considered alongside traffic exposure when analysing accident severity, as variations in collision frequency do not necessarily indicate differences in underlying risk.

Correlation analysis measures the strength and direction of the linear relationship between two variables. This visualisation explores the relationships between selected numerical variables that may influence road accident severity.

# Visualization 14: Geographical Distribution of Road Accidents

The geographical location of road traffic collisions provides important insights into the spatial distribution of accident occurrence across Great Britain.

This visualisation uses a hexbin density plot to display the geographical distribution of road traffic collisions across Great Britain. Hexagonal binning reduces overplotting and highlights areas with particularly high accident concentrations.

```python
# Create a copy of the dataset
map_df = df.copy()

# Remove records with missing coordinates
map_df = map_df.dropna(subset=['latitude', 'longitude'])

# Display first few coordinate values
map_df[['latitude', 'longitude']].head()
```
